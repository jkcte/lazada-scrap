```python
import json

with open('products.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
print(len(data))
```

```python
no_review_count = sum(1 for item in data if not item.get('reviews'))
print(no_review_count)
```

```python
filtered_data = [item for item in data if not item.get('reviews')]
with open('products_with_no_reviews.json', 'w', encoding='utf-8') as out_f:
    json.dump(filtered_data, out_f, ensure_ascii=False, indent=2)
```
