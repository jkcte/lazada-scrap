```python
import json

with open('products.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
print(len(data))
```

```python
no_review_count = sum(1 for item in data if not item.get('reviews'))
print(no_review_count)
```

```python
filtered_data = [item for item in data if not item.get('reviews')]
with open('products_with_no_reviews.json', 'w', encoding='utf-8') as out_f:
    json.dump(filtered_data, out_f, ensure_ascii=False, indent=2)
```

```python
import random
import json

half_len = len(filtered_data) // 2
half_data = filtered_data[:half_len]
random.shuffle(half_data)
split1 = half_data[:half_len // 2]
split2 = half_data[half_len // 2:]

with open('filtered_data_part1.json', 'w', encoding='utf-8') as f1:
    json.dump(split1, f1, ensure_ascii=False, indent=2)

with open('filtered_data_part2.json', 'w', encoding='utf-8') as f2:
    json.dump(split2, f2, ensure_ascii=False, indent=2)
```

```python
with open('filtered_data_part1.json', encoding='utf-8') as f:
    products = json.load(f)

print(len(products))
```

```python
filtered = sum(1 for item in products if item.get('reviews'))
print(filtered)
```

```python
with open('product_with_reviews.json', 'w', encoding='utf-8') as f:
    combined = [item for item in products if item.get('reviews')] + [item for item in data if item.get('reviews')]
    json.dump(combined, f, ensure_ascii=False, indent=2)
    print(len(combined))
```

```python
filtered_no_reviews = [item for item in products if not item.get('reviews')]
with open('filtered_data_part1_no_reviews.json', 'w', encoding='utf-8') as f:
    json.dump(filtered_no_reviews, f, ensure_ascii=False, indent=2)
    print(len(filtered_no_reviews))
```

```python
import json

with open('filtered_data_part1.json', 'r', encoding='utf-8') as f:
    with_filted = json.load(f)


filtered_no_reviews = [item for item in with_filted if not item.get('reviews')]

print(len(with_filted ) - len(filtered_no_reviews))
```

```python
import json

with open('filtered_data_part1.json', 'r', encoding='utf-8') as f:
    part1 = json.load(f)
with open('filtered_data_part2.json', 'r', encoding='utf-8') as f:
    part2 = json.load(f)
with open('products.json', 'r', encoding='utf-8') as f:
    orig = json.load(f)
with open('product_with_reviews.json', 'r', encoding='utf-8') as f:
    reviews = json.load(f)

combined = part1 + part2 + orig + reviews
filtered_combine_with_reviews = [item for item in combined if item.get('reviews')]
with open('filtered_data_combined_with_reviews.json', 'w', encoding='utf-8') as f:
    json.dump(filtered_combine_with_reviews, f, ensure_ascii=False, indent=2)

len(filtered_combine_with_reviews)
```

```python
reviews_count_per_product = [len(item.get('reviews', [])) for item in combined]
print(reviews_count_per_product)
count = 0
for review in reviews_count_per_product:
    if review > 0:
        count += review

print(count)
```
